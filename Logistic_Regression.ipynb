```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Importing Default Dataset
default_data = pd.read_excel('Finance_Default.xlsx', usecols=['default', 'student', 'balance', 'income'])
X = default_data.iloc[:, 1:]
y = default_data.iloc[:, 0]
```

```python
#Coding 'Yes', 'No' to 1, 0
X.loc[:, ['student']] = X['student'].apply(lambda x : 1 if x=='Yes' else 0)
y = y.apply(lambda x : 1 if x=='Yes' else 0)
```

```python
#Splitting data into train and test datasets
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.33, random_state=42)
```

```python
#Logistic Regression Regressor
class LogisticRegressor:
  def __init__(self):
    self.coef_ = None
    self.intercept_ = None

  def fit(self, X_train, y_train, eta = 0.1):
    n = X_train.shape[0]
    no_predictors = X_train.shape[1]
    X_train = np.insert(X_train,0, values=1, axis = 1)
    y_train = np.array(y_train).reshape(-1,1)    #Row vector
    theta = np.random.randn(no_predictors+1).reshape((no_predictors+1, 1))

    #Sigmoid function
    def sigmoid(z):
      return 1/ (1 + np.exp(-z))

    n_epochs = 10000
    for epoch in range(n_epochs):
      y_predicted = sigmoid(X_train.dot(theta))
      gradient_vector = (1/n)*(X_train.T.dot(y_predicted- y_train))
      theta = theta - eta* gradient_vector
    self.coef_ = theta[1:].reshape(1, -1)    #Column Vector
    self.intercept_ = theta[0]

  def predict(self, X_test):
    #Sigmoid function
    def sigmoid(z):
      return 1/ (1 + np.exp(-z))
    X_test = np.insert(X_test,0, values=1, axis = 1)
    theta = np.concatenate((self.intercept_, self.coef_), axis=None).reshape((-1,1))
    y_predicted = sigmoid(X_test.dot(theta))
    return y_predicted
```

```python
#Scaling X_train dataset before fitting
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_train)

#Finding best fit parameters using sklearn model
from sklearn.linear_model import LogisticRegression
sk_regressor = LogisticRegression(max_iter=100)
sk_regressor.fit(X_scaled, y_train)

#Finding best fit parameters using custom model
lrgd_regressor = LogisticRegressor()
lrgd_regressor.fit(X_scaled, y_train)

#Comparing closed form solution and gradient descent
print('Estimated parameters from sklearn regressor')
print('weights:', sk_regressor.coef_)
print('intercept:', sk_regressor.intercept_)
print('Estimated parameters from custom regressor')
print('weights:', lrgd_regressor.coef_)
print('intercept:', lrgd_regressor.intercept_)
```

```python
#Scaling X_train dataset before testing
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_test)
y_predicted = lrgd_regressor.predict(X_scaled)
y_predicted
```
